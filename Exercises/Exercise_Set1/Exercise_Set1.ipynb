### c)

Split the data into training and test sets with a 70%-30% ratio. Use `random_state=42` for reproducible results. Make sure that the class ratio is kept the same across the training and the test set. 

Note that in order to create the training and test sets, you first need to separate the features and the response into two variables, so you can use them as proper inputs.

**(10 points)**

```python
X = df.drop(columns="class")
y = df["class"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42, stratify=y)
```

### d)

(Standard) scale the training and test sets. **(15 points)**

```python
scale = StandardScaler()

X_train_scaled = scale.fit_transform(X_train)
X_test_scaled = scale.transform(X_test)
```

### e)

Create and train a Logistic Regression model. (You are expected to know which set to use to train the model.) Do **not** use any Lasso or Ridge penalty.

**(15 points)**

```python
model = LogisticRegression(penalty=None)

model.fit(X_train_scaled, y_train)
```

### f)

Evaluate the model by printing its accuracy, recall and confusion matrix. (You are expected to know which set to use to evaluate the model.) You can assume that the probability threshold is 0.5, which is the default value.

**(20 points)**

```python
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
matrix = confusion_matrix(y_test, y_pred)

print("Accuracy:", accuracy)
print("Recall:", recall)
matrix
```

With an accuracy of 98.8% and a recall of 99.5%, the model performs really well on the test set.

```python
# some extra visualization for myself

import matplotlib.pyplot as plt

scores = model.decision_function(X_test_scaled)

plt.scatter(range(len(scores)), scores, c=y_test)

plt.xlabel("Test Observation")
plt.ylabel("Decision Score")
plt.axhline()

plt.show()
```
