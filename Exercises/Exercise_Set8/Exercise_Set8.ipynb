```python
def update_step(X, cluster_preds):
    df = pd.DataFrame(np.column_stack((X, cluster_preds)))
    centroids = df.groupby(X.shape[1]).mean()
    return centroids.to_numpy()

def assignment_step(X, centroids):
    distances = np.linalg.norm(X[:, None, :] - centroids[None, :, :], axis=2)
    return np.argmin(distances, axis=1)
```
