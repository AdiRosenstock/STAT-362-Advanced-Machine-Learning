Import all the libraries and tools you need below.

```python
import tensorflow as tf
import numpy as np
import keras
from sklearn.metrics import accuracy_score, confusion_matrix
```

Run the line given below to read the MNIST dataset. Reshape the training and test features. (This should be ready from the previous exercise set.)

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.reshape(60000, 784)
x_test = x_test.reshape(10000, 784)
```

### 1)

Using a **keras tool**, one-hot-encode the training target/response values. **(10 points)**

```python
y_train_encoded = keras.utils.to_categorical(y_train)
```

### 2)

Explain why one-hot-encoding the target values is necessary for classification with a neural network. (We have only one-hot-encoded categorical features up until this point!) **(10 points)**

One hot encoding converts each target value into a vector of zeros and one 1 , with the 1 representing the correct class. This allows the neural network to compare its predicted probabilities for each class with the actual class when using categorical cross entropy. Also, it avoids treating the digits as numerical values with an ordering.

### 3)

Create a **five-layer** network. Use 200, 100 and 50 nodes for each hidden layer, respectively. **You need to use both `InputLayer` and `Dense` objects (and only them) for credit.** 

Add the proper non-linear functions to the hidden and output layers. For the hidden layers, you should use the **most common** function that avoids the vanishing gradients problem. (Use the actual function, not its modified versions. The extensions will come later.) For the output layer, you are expected to know the only proper function to use.

**In or between the layers, do not add any extra inputs that are not instructed. All components of a neural network will appear gradually in future exercise sets.**

**(30 points)**

```python
model = keras.Sequential([
    keras.layers.InputLayer(shape=(784,)),
    keras.layers.Dense(200, activation="relu"),
    keras.layers.Dense(100, activation="relu"),
    keras.layers.Dense(50, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])
```

### 4)

Why is it very common practice to have less nodes in deeper hidden layers? **(10 points)**

It is a common practice for deeper hidden layers to have fewer neurons because the network gradually learns more complex and useful features from the data. The earlier layers capture simpler patterns, while deeper layers combine them into more meaningful representations. Using fewer neurons also reduces the number of parameters and computational cost, and at the same time it can help prevent overfitting.

### 5)

`compile` the network **only** with the cost function. You need to use the proper cost function for the task at hand and specify it as a **string, not as an object**. Do not account for any sparsity in the data.

Print the network summary.

**(10 points)**

```python
model.compile(loss="categorical_crossentropy")

model.summary()
```

### 6)

Train the network for 5 epochs and a batch size of 100 (more on batch size later). **Do not use any other inputs.** Save the training line to an output variable, named `history`.

Print the test accuracy and the test confusion matrix. Note that you need to process the direct model output.

**(30 points)**

**You will fine-tune this network in the next exercise set!**

```python
history = model.fit(x_train, y_train_encoded, epochs=5, batch_size=100)

y_pred = model.predict(x_test)
y_pred_classes = np.argmax(y_pred, axis=1)

print("Test accuracy:", accuracy_score(y_test, y_pred_classes))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred_classes))
```
